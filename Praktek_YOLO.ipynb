```python
from IPython.display import HTML
from base64 import b64encode
import os

# Path input dan output video
save_path = 'runs/detect/train4/test1.mp4'
compressed_path = 'result_compressed.mp4'

# Pastikan file input ada
if os.path.exists(save_path):
    # Kompres video menggunakan ffmpeg
    os.system(f"ffmpeg -y -i {save_path} -vcodec libx264 -crf 28 {compressed_path}")

    # Baca dan encode video sebagai base64
    mp4 = open(compressed_path, 'rb').read()
    data_url = "data:video/mp4;base64," + b64encode(mp4).decode()

    # Tampilkan video di notebook
    display(HTML(f"""
    <video width=400 controls>
        <source src="{data_url}" type="video/mp4">
    </video>
    """))
else:
    print("File video input tidak ditemukan:", save_path)
```
